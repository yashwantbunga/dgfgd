Here are some examples to practice understanding the assignment operator:

1.  **Simple Assignment:**
    ```python
    a = 10
    b = 20
    c = a + b
    print(c) # Output: 30
    ```
    In this example, `=` assigns the value `10` to `a`, `20` to `b`, and the result of `a + b` (which is `30`) to `c`.

2.  **Reassignment:**
    ```python
    message = "Hello"
    print(message) # Output: Hello
    message = "World"
    print(message) # Output: World
    ```
    Here, the variable `message` is first assigned the string "Hello" and then reassigned to "World".

3.  **Compound Assignment Operators:**
    ```python
    count = 5
    count += 3 # Equivalent to: count = count + 3
    print(count) # Output: 8

    price = 100
    price *= 0.9 # Equivalent to: price = price * 0.9
    print(price) # Output: 90.0
    ```
    Compound assignment operators combine an arithmetic operation with an assignment. `+=` adds a value to the variable, and `*=` multiplies the variable by a value.

####**TASK P13.2** Explore file object functions to read and print practice.txt

````python
from google.colab import files


uploaded = files.upload()
filename = next(iter(uploaded))

print("\n--- Reading first line with readline() ---")
with open(filename, "r") as file:
    first_line = file.readline()
    print(first_line)

print("\n--- Reading all lines into a list with readlines() ---")
with open(filename, "r") as file:
    all_lines = file.readlines()
    # Print the first few lines to avoid flooding the output if the file is very large
    for i, line in enumerate(all_lines):
        if i < 5:
            print(line, end='') # end='' to prevent double newlines as lines already contain them
        else:
            break
    if len(all_lines) > 5:
        print("... (truncated, total lines: {}) ".format(len(all_lines)))
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
from google.colab import files
uploaded = files.upload()
filename = next(iter(uploaded))

# Read the words into a list.
with open(filename, "r") as data:
    words = data.read().split()

# Use count() to count each unique word.
word_counts = {}
for word in words:
    if word not in word_counts:
        word_counts[word] = words.count(word)

# Print each word and its count.
for word, count in word_counts.items():
    print(word, count)
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

####**TASK P23.1** Modify the script from TASK P22.1 so that it prints the GC content for each sequence in the file

````python
uploaded = files.upload()
filename = next(iter(uploaded))

# Calculate and print each sequence's ID and GC percentage.
with open(filename, "r") as data:
    for record in parse(data, "fasta"):
        sequence = str(record.seq).upper()
        gc_content = (sequence.count("G") + sequence.count("C")) / len(sequence) * 100
        print(record.id, f"{gc_content:.6f}")
````

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
uploaded = files.upload()
filename = next(iter(uploaded))

# Track the sequence with the highest GC content.
highest_gc = -1
highest_id = ""

with open(filename, "r") as data:
    for record in parse(data, "fasta"):
        sequence = str(record.seq).upper()
        gc_content = (sequence.count("G") + sequence.count("C")) / len(sequence) * 100

        if gc_content > highest_gc:
            highest_gc = gc_content
            highest_id = record.id

# Print the ID and GC percentage on separate lines.
print(highest_id)
print(f"{highest_gc:.6f}")
````
